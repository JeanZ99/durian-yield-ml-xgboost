# Exploratory Data Analysis
Durian Yield ML — Chanthaburi Province, Thailand


```python
import sys; sys.path.insert(0, '../src')
import pandas as pd, matplotlib.pyplot as plt, seaborn as sns
from config import DATA_PROC, TARGET, FEATURES_ALL, GROUP_CV
df = pd.read_parquet(DATA_PROC / 'durian_ml_processed.parquet')
print(df.shape)
df.describe().T
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
df[TARGET].hist(bins=40, ax=axes[0]); axes[0].set_title('Yield distribution (kg/ha)')
df.groupby(GROUP_CV)[TARGET].median().sort_values().plot.bar(ax=axes[1])
axes[1].set_title('Median yield by district'); plt.tight_layout(); plt.show()
```

```python
feats_avail = [f for f in FEATURES_ALL if f in df.columns][:20]
corr = df[feats_avail + [TARGET]].corr()
plt.figure(figsize=(14, 12))
sns.heatmap(corr, annot=False, cmap='coolwarm', center=0)
plt.title('Feature correlation matrix'); plt.tight_layout(); plt.show()
```
