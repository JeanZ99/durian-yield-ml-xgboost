# Model Diagnostics
Residual analysis, predicted vs. actual, and LODO breakdown.


```python
import sys; sys.path.insert(0, '../src')
import pandas as pd, numpy as np, xgboost as xgb, matplotlib.pyplot as plt
from config import DATA_PROC, MODELS_DIR, FEATURES_ALL, TARGET
df = pd.read_parquet(DATA_PROC / 'durian_ml_processed.parquet')
feats = [f for f in FEATURES_ALL if f in df.columns]
model = xgb.XGBRegressor(); model.load_model(str(MODELS_DIR / 'xgb_final.json'))
y_pred = model.predict(df[feats]); y_true = df[TARGET].values
residuals = y_true - y_pred
print(f'Full-data R2 = {1 - np.var(residuals)/np.var(y_true):.4f}')
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))
axes[0].scatter(y_pred, y_true, alpha=0.4, s=15)
lim = [min(y_pred.min(),y_true.min()), max(y_pred.max(),y_true.max())]
axes[0].plot(lim, lim, 'r--'); axes[0].set_xlabel('Predicted'); axes[0].set_ylabel('Actual')
axes[0].set_title('Predicted vs Actual')
axes[1].hist(residuals, bins=40, edgecolor='black')
axes[1].set_title('Residuals'); plt.tight_layout(); plt.show()
```
